***1.2. Use sns.histplot to visualize the distribution of house prices.***

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize = (10, 6))
sns.histplot(df['SalePrice'], kde = True, bins = 30)
plt.title('Distribution of SalePrice')
plt.xlabel('House price')
plt.ylabel('House count')
plt.show()
print("Min: ", df['SalePrice'].min())
print('Max: ', df['SalePrice'].max())
print('Mean: ', df['SalePrice'].mean()) #cộng lại tất cả giá tiền của căn nhà rồi chia số lượng căn nhà trong dataset. 
print('Median: ', df['SalePrice'].median()) #giá của căn nhà ở chính giữa dataset khi xếp theo giá tiền.
print('Standard Deviation: ', df['SalePrice'].std())
```

It is noticeable that the histogram is right-skewed (mean > median)

 ***1.3. Calculate the correlation between the features and SalePrice using a correlation matrix and visualize it with sns.heatmap***

```python
import seaborn as sns
import matplotlib.pyplot as plt

#calcute correction between numrical features and saleprice
numrical_features = df.select_dtypes(include = ['int64', 'float64'])
corr = numrical_features.corr()

top_corr = corr['SalePrice'].abs().sort_values(ascending= False).head(20)
print(top_corr)
# Vẽ heatmap cho ma trận tương quan
plt.figure(figsize=(15, 12))
sns.heatmap(corr,cmap='magma')
plt.show()
```

The correlation between SalePrice andOverallQual(đánh giá chất liệu và độ hoàn thiện tổng thể của nhà), GrLivArea(diện tích sinh hoạt trên mặt đất), GarageCars, GarageArea, TotalBsmtSF(tổng diện tích tầng hầm ), 1stFlrSF(DIEN TICH SAN TANG 1), FullBath(phòng tắn đầy đủ trên tầng trên )is greater than 0.5.

=> These features have a strong influence on house prices.

=> Most of these variables are related to the size or living area of the property, suggesting that larger houses generally sell for higher prices.

=> In addition, time-related factors may also play an important role in determining sale prices.

 ***1.4. Identify missing values and detect outliers to ensure data quality before model training.***

```python
#missing values
missing = df.isnull().sum() # return the number of missing values in each column => a series with index as column name and value as sum of missing value
missing = missing[missing > 0].sort_values(ascending=False)
missing_percentage = (missing / len(df))*100 
pd.DataFrame({'Missing cout': missing, 'Missing percentage': missing_percentage}) # merge two series into a dataframe with two columns
```

Many columns have a high percentage of missing values, but these values often indicate the absence of a feature rather than missing data. 

=>Therefore, the features are divided into three categories:
1. Nhóm điền "None" (cột dạng chữ — category)
